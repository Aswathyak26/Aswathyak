```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = data1.drop('Price', axis = 1) #independent data
y= data1['Price'] #dependent data
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

### Linear Regression

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg = lin_reg.fit(x_train,y_train)
predictions_linear = lin_reg.predict(x_test)
from sklearn.metrics import mean_squared_error, r2_score
print('LINEAR REGRESSION')
mse_linear = mean_squared_error(y_test, predictions_linear)
r2_linear = r2_score(y_test, predictions_linear)
print('Mean Squared Error = ',mse_linear)
print('R Squared Error = ',r2_linear)
```

```python
value_linear = lin_reg.predict(pd.DataFrame({'Year':[2022]}))
print('Predicted price for 2022 model using Liner Regression = ', value_linear[0])
```

### Lasso Regression

```python
from sklearn.linear_model import Lasso
lasso_model = Lasso(alpha = 0.1)
lasso_model = lasso_model.fit(x_train,y_train)
predictions_lasso = lasso_model.predict(x_test)
from sklearn.metrics import mean_squared_error, r2_score
print('\nLASSO REGRESSION')
mse_lasso = mean_squared_error(y_test, predictions_lasso)
r2_lasso = r2_score(y_test,predictions_lasso)
print('Mean Squared Error = ',mse_lasso)
print('R Squared Error = ',r2_lasso)
```

```python
value_lasso = lasso_model.predict(pd.DataFrame({'Year':[2022]}))
print('Predicted price for 2022 model using Lasso Regression = ', value_lasso[0])
```

```python
comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Lasso Regression'],
    'MSE': [mse_linear, mse_lasso],
    'R2 Score': [r2_linear, r2_lasso]
})

print(comparison)
```

```python
# Lasso Regression has slightly lower MSE #lower MSE means fewer prediction mistakes.
# Lasso Regression has slightly higher R2 Score #R² value ranges from 0 to 1. Closer to 1 means better model.
```

```python
### Lasso Regression showed slightly better performance than Linear Regression, although both models produced nearly identical results ,
### indicating that the dataset has a simple linear relationship between year and price.
#Since this dataset contains only one feature (Year), both Linear and Lasso Regression produced almost identical results.
```
