```python
#dataset of 10 sales records
orders = [
    {"order_id": 1, "customer": "Aswathy", "item": "Notebook", "quantity": 5, "price_per_item": 35.0, "order_date": "2026-04-07"},
    {"order_id": 2, "customer": "Arya", "item": "Pencil", "quantity": 10, "price_per_item": 10.0, "order_date": "2026-04-08"},
    {"order_id": 3, "customer": "Arya", "item": "Pen", "quantity": 10, "price_per_item": 10.0, "order_date": "2026-04-09"},
    {"order_id": 4, "customer": "Aswathy", "item": "Chocolate", "quantity": 3, "price_per_item": 5.0, "order_date": "2026-04-10"},
    {"order_id": 5, "customer": "Arya", "item": "Shirt", "quantity": 2, "price_per_item": 800.0, "order_date": "2026-04-11"},
    {"order_id": 6, "customer": "Siju", "item": "Pant", "quantity": 2, "price_per_item": 1200.0, "order_date": "2026-04-12"},
    {"order_id": 7, "customer": "Anju", "item": "Saree", "quantity": 3, "price_per_item":999.0, "order_date": "2026-04-13"},
    {"order_id": 8, "customer": "Ishaan", "item": "Ball", "quantity": 1, "price_per_item": 350.0, "order_date": "2026-04-14"},
    {"order_id": 9, "customer": "Ishaan", "item": "Teddy Bear", "quantity": 2, "price_per_item": 150.0, "order_date": "2026-04-15"},
    {"order_id": 10, "customer": "Aswathy", "item": "Radio", "quantity": 1, "price_per_item": 1500.0, "order_date": "2026-04-16"}
]

#Print out the dataset
print("Dataset:\n")
for order in orders:
    print(order)

#Calculate the total revenue
total_rev = 0
for order in orders:
    total_rev += order["quantity"] * order["price_per_item"]
print("Total Revenue is :", total_rev)

#total order value exceeds ₹2,000
high_order = []
for order in orders:
    value = order["quantity"] * order["price_per_item"]
    if value > 2000:
        high_order.append(order)
print("\nOrder value exceeds ₹2,000 are:")
for i in high_order:
    print(i)

#item with the highest total quantity sold
item_count = {}
for order in orders:
    item = order["item"]
    if item in item_count:
        item_count[item] += order["quantity"]
    else:
        item_count[item] = order["quantity"]
#max_item = max(item_count, key=item_count.get) will not give result correctly if 2 items have same number
max_quantity = max(item_count.values())
max_items = [item for item, qty in item_count.items() if qty == max_quantity]

print("\nHighest sold item(s):", max_items, "| Quantity:", max_quantity)

#function apply_discount
def apply_discount(orders, discount_rate):
    new_orders = []
    for order in orders:
        new_order = order.copy()
        new_order["price_per_item"] = new_order["price_per_item"] * (1 - discount_rate)
        new_orders.append(new_order)
    return new_orders
print("After Discount :")
discounted = apply_discount(orders, 0.10)
for i in discounted:
    print(i)

# Local vs Global Scope Demonstration
store_name = "ShopEase"
def demo():
    local_msg = "This is a local variable"
    print("\n Inside function:")
    print("Store Name (Global):", store_name)
    print("Local Message:", local_msg)
demo()

print("\nOutside function:")
print("Store Name (Global):", store_name)
#print(local_msg)  # This would cause an error (local variable)


#custom iterator class
class OrderIterator:
    def __init__(self, orders):
        self.orders = orders
        self.i = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.i < len(self.orders):
            result = self.orders[self.i]
            self.i += 1
            return result
        else:
            raise StopIteration

print(" \nCustom Iterator:")
for order in OrderIterator(orders):
    print(order)

#grouping all orders
grouped = {
    customer: [order for order in orders if order["customer"] == customer]
    for customer in set(order["customer"] for order in orders)
}

print("\nGrouped data is:")

for customer, orders_list in grouped.items():
    print(f"\n{customer}:")
    for order in orders_list:
        print(order)
```
