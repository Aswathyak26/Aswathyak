### CHECK AND REMOVE DUPLICATES

```python
# Duplicate Rows
print("Duplicate Rows:", data1.duplicated().sum())
# Remove duplicate rows
data1 = data1.drop_duplicates()
```

```python
# Duplicate Columns
duplicate_columns = data1.columns[data1.T.duplicated()]
print("Duplicate Columns:")
print(duplicate_columns)
# Remove duplicate columns
data1 = data1.loc[:, ~data1.T.duplicated()]
```

### CHECK MISSING VALUES

```python
print("\nMissing Values :")
data1.isnull().sum()
```

```python
# Drop unwanted column
df1= data1.drop(columns=['No of Times Visited']) #This column is removed because out of 21613 entries, 19489 is missing.
df1 = df1.dropna(subset=['Sale Price']) # Removed rows where target variable is missing
```

```python
df1.isnull().sum()
```

### HANDLING MISSING VALUES

```python
# Missing numerical columns
missing_num_cols = [col for col in df1.select_dtypes(include=['number']).columns if df1[col].isnull().sum() > 0]

print("Missing Numerical Columns:\n", missing_num_cols)
```

```python
df1[missing_num_cols].hist(figsize=[15, 10], bins=20)
plt.show()
```

```python
# 1.  Missing Continuous numerical features - Median 
# Median is used because these features are skewed
# Missing Sale Price values - we removed rows where target variable is missing
median_cols = ['No of Bathrooms','Flat Area (in Sqft)','Lot Area (in Sqft)','Area of the House from Basement (in Sqft)','Living Area after Renovation (in Sqft)']
for i in median_cols:
    df1[i] = df1[i].fillna(df1[i].median())
```

```python
# Location-based features - Mode 
# Mode is used because these columns represent geographical/location information. Using median will give false latitude and longitude.
mode_cols = ['Zipcode', 'Latitude', 'Longitude']
for i in mode_cols:
    df1[i] = df1[i].fillna(df1[i].mode()[0])
```

```python
#2. Missing categorical features
# Categorical missing values filled using mode because mode represents the most frequent category
missing_cat_cols = [col for col in df1.select_dtypes(include=['object']).columns 
                   if df1[col].isnull().sum() > 0]
print("\nMissing Categorical Columns:\n", missing_cat_cols) 
```

###  OUTLIER DETECTION USING BOXPLOT

```python
# The below columns were selected for outlier analysis because they are continuous numerical features related to house size, area and structure.
# Other columns are avoided because:
# Sale Price is the target variable
# No of Bedrooms, No of Bathrooms, No of Floors, Overall Grade are discrete/count or rating variables
# Waterfront View, Condition of the House are categorical variables
# ID is identifier column
# Age of House (in Years), Renovated Year, Zipcode, Latitude, Longitude, Date House was Sold - these are not suitable for IQR-based outlier removal.
columns_to_clean_outliers = ['Flat Area (in Sqft)', 
    'Lot Area (in Sqft)', 'Area of the House from Basement (in Sqft)', 
    'Basement Area (in Sqft)', 'Living Area after Renovation (in Sqft)', 
    'Lot Area after Renovation (in Sqft)']
```

```python

fig, axes = plt.subplots(3, 2, figsize=(20, 16))
axes = axes.flatten()
i = 0 # Start a manual counter at index 0
for col in columns_to_clean_outliers:
    axes[i].boxplot(df1[col])
    axes[i].set_title(col)
    i = i + 1  # Manually move to the next subplot position for the next column
plt.tight_layout()
plt.show()
```

###  REMOVE OUTLIERS USING IQR METHOD

```python
# Based on boxplot analysis, significant outliers were observed mainly in lot-area related features.
# Therefore, IQR-based outlier treatment was applied only to those columns.
outliers = ['Lot Area (in Sqft)','Lot Area after Renovation (in Sqft)']  #This can be also log transformed to avoid potential data loss.
for col in outliers:
    Q1 = np.percentile(df1[col],25)
    Q3 = np.percentile(df1[col],75)
    IQR = Q3 - Q1
    lwr_lim = Q1 - 1.5 * IQR
    upr_lim = Q3 + 1.5 * IQR
    df1 = df1[(df1[col] >= lwr_lim ) & (df1[col] <= upr_lim)] # This drops the entire row if a feature is an outlier
print("\nShape After Outlier Removal:", df1.shape)
```

### IDENTIFY NUMERICAL & CATEGORICAL COLUMNS

```python
numerical_cols = df1.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = df1.select_dtypes(include=['object']).columns

print('Numerical Columns:\n', numerical_cols)
print('\nCategorical Columns:\n', categorical_cols)
```

### ENCODING CATEGORICAL VARIABLES

```python
df1['Condition of the House'].unique()
```

```python
df1['Waterfront View'].unique()
```

```python
df1['Date House was Sold'].unique()
```

```python
# Nominal category - One-Hot Encoding
df1 = pd.get_dummies(df1, columns=['Waterfront View'], drop_first=True, dtype=int) #drop_first=True drops 'Waterfront View_No' to prevent redundancy

# Ordinal category - Label Encoding
condition_mapping = {'Bad': 1,'Okay':2, 'Fair': 3, 'Good': 4, 'Excellent': 5} 
df1['Condition of the House'] = df1['Condition of the House'].map(condition_mapping)
```

```python
# Convert the date column into a true Python datetime format
df1['Date House was Sold'] = pd.to_datetime(df1['Date House was Sold'], format='%d %B %Y',errors='coerce')

# Extract the Year,Month and day of week
df1['Year of Sale'] = df1['Date House was Sold'].dt.year
df1['Month of Sale'] = df1['Date House was Sold'].dt.month
df1['Day of Week of Sale'] = df1['Date House was Sold'].dt.dayofweek  # This gives 0=Monday,.. 6=Sunday

# Drop the original date column since we don't need it anymore
df1 = df1.drop(columns=['Date House was Sold'])
```

```python
df1.head()
```

### SCALING NUMERICAL VARIABLES

```python
#Below columns are avoided because:
# ID is not physical numerical measurement
#Zipcode is location identifier and excluded from scaling
# Sale Price is the target variable 
# Waterfront View_Yes - is binary encoded column 
#Condition of the House - Label Encoded
from sklearn.preprocessing import StandardScaler
scale_cols = []
all_numeric_columns = df1.select_dtypes(include=['number']).columns

for i in all_numeric_columns:
    if i not in ['ID', 'Zipcode', 'Sale Price','Waterfront View_Yes','Condition of the House']: 
        scale_cols.append(i)
        
std_scaler = StandardScaler()
df1[scale_cols] = std_scaler.fit_transform(df1[scale_cols])
print('Scaling complete. Output object type: ',type(df1))
```
